```python
import pandas as pd
import logging
from pathlib import Path
```

```python

log_dir = Path("../logs")
log_dir.mkdir(parents=True, exist_ok=True)

# Setup logging
logging.basicConfig(
    filename=log_dir / "data_transformation_log.txt",
    level=logging.INFO,
    format="%(asctime)s - %(levelname)s - %(message)s"
)
```

```python
# Log notebook start
logging.info("Starting data exploration for LinkedIn and Website datasets")
```

```python
# Load datasets
try:
    linkedin_df = pd.read_excel("../data/raw/linkedin_stats.xls", header=1)
    website_df = pd.read_csv("../data/raw/website_requests.csv")
    logging.info("Successfully loaded linkedin_stats.xls and website_requests.csv")
except Exception as e:
    logging.error(f"Error loading datasets: {str(e)}")
    raise
```

```python
# Display basic info
print("LinkedIn Dataset Info:")
print(linkedin_df.info())
print("\nWebsite Requests Dataset Info:")
print(website_df.info())
```

```python
# Check for missing values
print("\nMissing Values in LinkedIn Dataset:")
print(linkedin_df.isnull().sum())
print("\nMissing Values in Website Requests Dataset:")
print(website_df.isnull().sum())
```

```python
# Check for duplicates
linkedin_duplicates = linkedin_df.duplicated().sum()
website_duplicates = website_df.duplicated().sum()
print(f"\nLinkedIn Duplicates: {linkedin_duplicates}")
print(f"Website Duplicates: {website_duplicates}")
logging.info(f"Found {linkedin_duplicates} duplicates in LinkedIn dataset")
logging.info(f"Found {website_duplicates} duplicates in Website dataset")
```

```python
# Save sample data for reference
linkedin_df.head().to_csv("../data/processed/linkedin_sample.csv", index=False)
website_df.head().to_csv("../data/processed/website_sample.csv", index=False)
logging.info("Saved sample data for both datasets")
```

```python
# Convert LinkedIn date column
try:
    linkedin_df["Date"] = pd.to_datetime(linkedin_df["Date"], errors="coerce")
    logging.info(f"Converted LinkedIn 'Date' to datetime, {linkedin_df['Date'].isnull().sum()} nulls created")
except Exception as e:
    logging.error(f"Error converting LinkedIn 'Date': {str(e)}")
```

```python
# Convert numerical columns
numerical_columns = [
    "Impressions (organic)", "Impressions (sponsored)", "Impressions (total)",
    "Unique impressions (organic)", "Clicks (organic)", "Clicks (sponsored)",
    "Clicks (total)", "Reactions (organic)", "Reactions (sponsored)",
    "Reactions (total)", "Comments (organic)", "Comments (sponsored)",
    "Comments (total)", "Reposts (organic)", "Reposts (sponsored)",
    "Reposts (total)", "Engagement rate (organic)",
    "Engagement rate (sponsored)", "Engagement rate (total)"
]
```

```python
for col in numerical_columns:
    try:
        linkedin_df[col] = pd.to_numeric(linkedin_df[col], errors="coerce")
        logging.info(f"Converted LinkedIn column '{col}' to numeric, {linkedin_df[col].isnull().sum()} nulls created")
    except Exception as e:
        logging.error(f"Error converting LinkedIn column '{col}' to numeric: {str(e)}")
```
