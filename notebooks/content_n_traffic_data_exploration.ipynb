```python
import pandas as pd
import logging
from pathlib import Path
```

```python
# Setup logging
log_dir = Path("../logs")
log_dir.mkdir(exist_ok=True)
logging.basicConfig(
    filename=log_dir / "data_transformation_log_2.txt",
    level=logging.INFO,
    format="%(asctime)s - %(levelname)s - %(message)s"
)
```

```python
# Log notebook start
logging.info("Starting exploration of new content and traffic datasets")
```

```python
# Folders and files
folders = ['Content', 'Traffic']
files = ['Chart data.csv', 'Totals.csv', 'Table data.csv']
```

```python
# Load and inspect each file
for folder in folders:
    for file in files:
        path = Path(f"../data/raw/{folder}/{file}")
        try:
            df = pd.read_csv(path)
            logging.info(f"Successfully loaded {folder}/{file}")
        except Exception as e:
            logging.error(f"Error loading {folder}/{file}: {str(e)}")
            continue
        
        print(f"\n--- {folder}/{file} ---")
        print("First 5 Rows:")
        print(df.head())
        print("\nData Types:")
        print(df.dtypes)
        print("\nInfo:")
        print(df.info())
        print("\nMissing Values:")
        print(df.isnull().sum())
        print("\nDuplicates:")
        print(df.duplicated().sum())
        logging.info(f"Explored {folder}/{file}: {len(df)} rows, {len(df.columns)} columns")
```
