```python
# Log inspection start
logging.info("Inspecting cleaned content and traffic datasets")

# Files to verify
content_files = ["content_chart_data_cleaned.parquet", "content_totals_cleaned.parquet", 
                 "content_table_data_cleaned.parquet", "content_table_total.parquet"]
traffic_files = ["traffic_chart_data_cleaned.parquet", "traffic_totals_cleaned.parquet", 
                "traffic_table_data_cleaned.parquet", "traffic_table_total.parquet"]
```

```python
for file in content_files:
    try:
        df = pd.read_parquet(f"../data/processed/{file}")
        print(f"\n--- {file} ---")
        print("First 5 Rows:")
        print(df.head())
        print("\nData Types:")
        print(df.dtypes)
        print("\nMissing Values:")
        print(df.isnull().sum())
        print("\nDuplicates:")
        print(df.duplicated().sum())
        logging.info(f"Inspected {file}: {len(df)} rows, {len(df.columns)} columns")
    except Exception as e:
        logging.error(f"Error loading {file}: {str(e)}")
```
